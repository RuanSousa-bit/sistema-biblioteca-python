```python
class Livro:
    def __init__(self, titulo, autor, ISBN, disponibilidade="Disponível"):
        self.titulo = titulo
        self.autor = autor
        self.ISBN = ISBN
        self.disponibilidade = disponibilidade

    def emprestar(self):
        if self.disponibilidade == "Disponível":
            self.disponibilidade = "Emprestado"
            return "Livro emprestado"
        return "Livro indisponível"

    def devolver(self):
        self.disponibilidade = "Disponível"
        return "Livro devolvido"

class Usuario:
    def __init__(self, nome, ID, historico_de_emprestimo="Sem empréstimos"):
        self.nome = nome
        self.ID = ID
        self.historico_de_emprestimo = historico_de_emprestimo

class Biblioteca:
    def __init__(self, nome, livro, usuario):
        self.nome = nome
        self.livro = livro
        self.usuario = usuario

    def emprestar_livro(self):
        resultado = self.livro.emprestar()
        if resultado == "Livro emprestado":
            self.usuario.historico_de_emprestimo = self.livro.titulo
        print(f"{self.usuario.nome} tentou emprestar '{self.livro.titulo}': {resultado}")

    def devolver_livro(self):
        print(f"{self.usuario.nome} devolveu '{self.livro.titulo}': {self.livro.devolver()}")


livro1 = Livro("1984", "George Orwell", "123456789")
usuario1 = Usuario("Carlos", 1)


biblioteca = Biblioteca("Biblioteca Central", livro1, usuario1)
biblioteca.emprestar_livro()
biblioteca.devolver_livro()
```
